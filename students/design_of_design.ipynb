## (8) Rationalism versus Empiricism in Design

Dos enfoques filosóficos en el proceso de diseño:
- **Racionalismo**: Los seres humanos son inherentemente sano, con suficiente pensamiento y educación, el diseñador puede, en teoría, formular un diseño perfecto sin necesidad de recurrir a la experimentación.

- **Empirismo**: Los seres humanos son inherentemente defectuosos, y por ello el proceso de diseño debe incluir iteración, pruebas y validación empírica para detectar y corregir errores.

<img src="../images/RacVSEmp.png" alt="rac_vs_emp" width="30%">

## (9) User Models — Better Wrong than Vague

*Modelo de Usuario:* representacion conpectual de cómo el diseñador imagina que el usuario piensa y utiliza su diseño.

Al diseñar:
- Documentar lo que saben sobre los usuarios: Qué espera lograr el usuario con el sistema? Sabe usar o ha usado sistemas similares? Qué dificultades puede tener?
- Solo los **mas experimentados**, también anotan lo que no saben pero asumen de los usuarios

## ¿Por qué es importante?

- Diseños en equipo
- Diseños complejos
- No diseñar para sí mismos, en lugar de para los usuarios

## ¿Qué ocurre en la práctica?

- Comienza el diseño
- Se definen modelos
- Se descubren cuánto desconocen.
- Esto genera preguntas
    - Se responden
    - Se realizan conjeturas

Brooks dice que, es mejor **hacer suposiciones explícitas (aunque sean incorrectas) que dejar vacíos en el modelo.**

Los diseñadores toman decisiones basadas en suposiciones de cualquier manera.

**Suposiciones erróneas pueden ser cuestionadas y corregidas; las imprecisas no.**

## (10) Inches, Ounces, Bits, Dollars—The Budgeted Resource

- **Recurso Presupuestado:** recurso escaso que restringe el diseño. No siempre es dinero: metros cuadrados, bits de memoria, toneladas de carga útil, tiempo (días, meses). Es importante gestionar este recurso. El recurso presupuestado **condiciona el diseño**.

- **Uso de Surrogados (o sustitutos):** a veces se utilizan sustitutos para representar el recurso excaso. Por ejemplo, un arquitecto puede usar los metros cuadrados del terreno, en lugar del presupuesto.

- **Cambio del Recurso Crítico:** a veces el recurso limitante se transforme durante el proceso de diseño, debido a condiciones del proyecto o avances tecnologicos. Ejemplo de OS/360, inicialmente la memoria era el recurso crítico, pero luego de mejoras en el acceso a disco, se pasó a gestionar las operaciones de disco como recurso limitante.

- **Gestión y Control del Recurso:** para mantener la integridad conceptual del diseño, es crucial:
    - Identificar explícitamente cuál es el recurso crítico.
    - Hacer un seguimiento público del recurso.
    - Controlarlo, asignando a una sola persona la responsabilidad de su administración.

## Diseño General vs Especial

- **Diseño General**:  
  Requiere equilibrar múltiples aplicaciones y modelos de usuario, lo que implica un análisis exhaustivo para lograr la integridad conceptual.
  
- **Diseño Especial**:  
  Al tener un propósito más restringido, el espacio de diseño ya está parcialmente definido, facilitando la obtención de resultados excelentes.

- **Ejemplo:** diseñar una habitación es más sencillo que diseñar un espacio público multifuncional.

Brook dice que, si se tiene un proyecto parece **carecer de restricciones**, es útil hcaer un analisis mas profundo de los objetivos y modelos de usuario para **identificar limitaciones**, pueden mejorar tanto el proceso creativo como el resultado final.

# 🏗️ A Computer Scientist’s Dream System for Designing Houses—Mind to Machine

Supongamos que uno pudiera imaginar la existencia de un sistema informático ideal para el diseño arquitectónico de casas y otros edificios. ¿Cómo sería?

- Inspirado en la evolución del sistema GRIP Molecular Graphics y en la interacción con sistemas gráficos avanzados.  
- Diseñado para ofrecer una **interfaz revolucionaria**, optimizada para arquitectos y diseñadores.  
- Su propósito: **conectar la mente del diseñador con la máquina**, permitiendo una experiencia fluida, intuitiva y natural.  

<img src="../images/duckarc.png" alt="duckarc" width="30%">


## 🎨 Principios de diseño – "Progressive Truthfulness"  

Una metodología de diseño basada en refinamiento progresivo:  

Una metodología que optimiza el proceso de diseño a través de refinamiento progresivo:  

- 🔄 **Mejorar en lugar de empezar desde cero**: Se parte de modelos existentes y se perfeccionan iterativamente.  
- ✅ **Coherencia garantizada**: Se mantiene la precisión del modelo en cada ajuste, evitando inconsistencias.  
- 🎯 **Ajustes incrementales**: Cada modificación acerca el diseño a la visión final del diseñador.  


**Ejemplo de interacción con el sistema:**
1. “Genera una casa georgiana de tres habitaciones.”
2. “Oriéntala hacia el norte.”
3. “Invierte el diseño izquierda-derecha.”
4. “Haz la sala de estar 14 pies de ancho.”
5. “Reduce la profundidad de la cocina un pie.”
6. “Cambia el exterior a blanco en lugar de ladrillo.”


## 📁 Biblioteca de Modelos y su Organización  

El sistema incluiría una **biblioteca extensa de modelos arquitectónicos** predefinidos, funcionando como punto de partida flexible para el diseño.  

🔹 **Características clave**:  
- 📂 *Estructura jerárquica* para una búsqueda rápida e intuitiva.  
- 📈 *Evolución constante*: Crece con el uso, permitiendo a los usuarios agregar sus propios modelos.  
- 🏠 *Escaneo y modelado 3D* para digitalizar casas existentes y facilitar remodelaciones.  
 
> La calidad de esta biblioteca es clave para evitar que el sistema limite la creatividad del diseñador.

## 🎨 Interfaz de Usuario y Métodos de Entrada  

El sistema está diseñado para permitir que el diseñador **exprese sus ideas de manera natural**, combinando múltiples formas de interacción.  

🔹 **Principios clave**:  
- *Entrada multimodal*: Uso de voz, gestos y herramientas digitales.  
- *Ritmo "Sustantivo-Verbo"*: Patrones de comandos intuitivos para mayor fluidez.  
    - Estructura: Verbo → Objeto → Adjetivo/Adverbio  
    - Ejemplo: "Haz esta pared de color verde." 

🔹 **Métodos de selección y modificación**: 
- 🎙️ *Comandos de voz* → "Selecciona la ventana izquierda."
- 🖱️ *Selección por apuntado* (2D y 3D).  
- 🔢 *Entrada numérica precisa* → "Puerta de 32 pulgadas de ancho." 
- ✏️ *Dibujo a mano alzada* (Sketching 2D y 3D con lápices digitales).

## ⚠️ Posibles Problemas del Enfoque *"Progressive Truthfulness"*  

1. **Limitación de la Creatividad**  
- ¿Los modelos predefinidos restringen la originalidad?  
- Los grandes diseñadores siempre han aprendido de precedentes, pero ¿puede esto frenar la innovación?

2. **Dependencia de la Biblioteca**  
- La calidad y cantidad de modelos predefinidos afecta directamente la *diversidad del diseño*.  
- Un sistema limitado en opciones podría resultar restrictivo para la creatividad.

3. **Complejidad de la Interacción**  
- Integrar voz, gestos y apuntado puede requerir un aprendizaje progresivo.  
- El sistema podría resultar complejo para usuarios sin experiencia previa.

## 🚫 Los Procesos Bloquean la Innovación  

Los **procesos de diseño corporativos** buscan orden, pero esto puede inhibir la innovación:

- ⚖️ Son **conservadores**: Buscan **predictibilidad** y basan decisiones en el pasado.  
- ❌ Funcionan como **sistemas de veto**, donde cada aprobación reduce la innovación.  
- 🏛️ Crean **Burocracia**

**Ejemplo en IBM**  
Durante el desarrollo del System/360 Model 20, el equipo en Alemania seguía meticulosamente el *proceso corporativo*, pero nunca lograba que sus productos fueran aceptados.

Un nuevo gerente con experiencia en "*evadir burocracia*" hizo los ajustes necesarios, y el producto se convirtió en un éxito.

> La creatividad necesita libertad para florecer.  


## 🤔 ¿Son los Procesos Necesarios?  

- **Sí**, pero en la fase correcta:
    - 💡 **Innovación disruptiva** → Se debe evitar la burocracia.  
    - 🔧 **Refinamiento y mejoras** → Aquí los procesos son útiles.

**Ejemplo:**  
- 📱 El primer iPhone fue una *innovación disruptiva*. 
- 📱 Los modelos siguientes fueron *mejoras progresivas* basadas en feedback del mercado.

> Los procesos son útiles para mejorar productos existentes, pero deben ser minimizados en las fases de innovación radical.



## 💻 Caso OS/360 - Sistema Operativo de IBM  

**Objetivo:**  
Crear un sistema operativo **robusto** y **compatible** con las computadoras **System/360**, que soporte multitarea y almacenamiento en disco.

 **Decisiones Clave**  
- 🛠️ *Sistema modular*: Diferentes configuraciones para hardware de gama baja, media y alta.  
- 🔌 *Soporte de dispositivos*: Sistema de control centralizado para varios dispositivos de entrada y salida.

 **Desafíos**  
- ⚠️ *Sobrecostos y retrasos* debido a la magnitud del proyecto.  
- 🐞 *Complejidad técnica* en la depuración del sistema.  
- 🧑‍💻 *Falta de experiencia en gestión de proyectos grandes*.

> OS/360 fue un hito en la computación, pero sus errores de planificación lo convirtieron en un caso de estudio sobre cómo no gestionar proyectos de gran escala.  


## 💻 Caso System/360 - Arquitectura de Computadoras

**Objetivo:**  
Diseñar una arquitectura **escalable** y **versátil** para diferentes niveles de procesamiento, desde pequeñas empresas hasta grandes centros de datos.

**Decisiones Clave**  
- 🖥️ Arquitectura de 8 bits por palabra y direcciones de 24 bits.  
- 🔄 *Conjunto de instrucciones unificado* para diversas configuraciones de hardware.  
- 🔌 *Interfaz estandarizada* de entrada/salida para conectar múltiples dispositivos.

**Desafíos**  
- ⚠️ *Compatibilidad futura* con direcciones de 32 bits no prevista, causando problemas de expansión.  
- 🧠 La *administración de memoria* requirió innovaciones en hardware y software.


> System/360 fue un gran éxito, pero algunas decisiones iniciales requirieron ajustes posteriores para seguir siendo competitivos.  


## 📚 Caso Diseño del Libro "Computer Architecture: Concepts and Evolution"

**Objetivo:**  
Escribir un libro que combinara teoría, estudios de caso y ejercicios prácticos sobre arquitectura de computadoras, dirigido a estudiantes y profesionales.

**Decisiones Clave**  
- 🧠 *Marco conceptual estructurado* para organizar los temas de forma clara.  
- 🖥️ 30 estudios de caso de arquitecturas para ilustrar la evolución de los conceptos en computación.  


**Desafíos**  
- ⚙️ *Precisión técnica*: Investigación y redacción largas debido a la complejidad del tema.  
- ⏳ *Evolución tecnológica rápida*: Algunos temas quedaron obsoletos antes de la publicación.  
- 🎓 *Adopción limitada*: Su enfoque técnico y nivel de detalle dificultaron su aceptación en programas educativos.

> El libro fue un referente en arquitectura de computadoras, pero su impacto limitado en la educación formal se debió a su enfoque técnico y denso.
