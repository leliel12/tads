<h1 style="color:#66bb6a; font-size:36px;">¿Que son los Bad Smells?</h1>

Los Bad Smells dentro del desarrollo de software son todos aquellos indicadores de que "algo está mal".

image.png

<h1 style="color:#66bb6a; font-size:36px;">¿Qué es un Code Smell?</h1>

Un Code Smell es una indicación superficial en el código fuente que usualmente corresponde a un problema más profundo en la arquitectura del sistema, la estructura del proyecto o la calidad del código en general.

Sin título.png

<h1 style="color:#66bb6a; font-size:36px;">¿Qué problemas puede causar un Code Smell?</h1>
<!-- Existen varios problemas por los cuales pueden aparecer los Code Smell -->

1. **Poca experiencia de los desarrolladores**

2. **Poco interés y desconocimiento**: Según encuestas de 2013, sabemos que el 32 % nunca ha oído hablar de ellos, y otro 50 % admite haberlos oído, pero nunca se ha interesado

3. **Alta presión**: Si el desarrollador debe entregar sus soluciones en un plazo limitado, a menudo debe tomar atajos que, consciente o inconscientemente , generan errores de código.

4. **Estado emocional y mental**

<h1 style="color:#66bb6a; font-size:36px;">¿Como solucionar los code smell?</h1>

<h1 style="color:#66bb6a; font-size:36px;">Refactorización</h1>

Es el proceso de modificar la **estructura interna** del código sin cambiar su **comportamiento observable**.

Apunta a mejorar la calidad del diseño plasmado en el codigo al alinear el código con **estandares**, **principios** y **criterios** de diseño como SOLID, minima sorpresa, reducir el acoplamiento y aumentar la coheion,....

Su objetivo es aumentar la claridad, mantenibilidad, testabilidad y extensibilidad del sistema.

> **Principio de mínima sorpresa**: Este establece que el comportamiento de un sistema, módulo, función o fragmento de código debe ser el que menos sorprenda (es decir, el mas intuitivo) a quien lo usa o lo lee.

<h2 style="color:#66bb6a; font-size:30px;">¿Cuándo conviene hacerla?</h2>

En orden de prioridad, es recomendable refactorizar:

1. **Antes de integrar cambios en ramas compartidas**, para evitar propagar diseño deficiente o difícil de mantener.

2. **Antes de un code review**, con el fin de presentar un código más limpio, coherente y legible.

3. **Antes de agregar una nueva funcionalidad**, cuando el diseño actual dificulta su incorporación.

4. **Al detectar code smells** (malos olores de diseño), como métodos largos, nombres ambiguos o bajo nivel de cohesión.

5. **Durante el mantenimiento del sistema**: al corregir bugs, adaptar componentes, o ejecutar tareas técnicas (sprints de refactorización, reducción de deuda técnica, etc.).

<h2 style="color:#66bb6a; font-size:30px;">¿Qué tener en cuenta al refactorizar?</h2>

Ademas de lo ya nombrado, es importante tener en cuenta:

**No modifica ni agrega funcionalidades**: Exclusivamente mejora la el diseño plasmado en el código para facilitar su evolución y mantenimiento.

**Control de cambios (versionado)**: utilizar control de versiones (como Git) para probar, comparar y, si es necesario, revertir refactorizaciones.

**Pequeños pasos**: cada refactorización debe ser mínima, **reversible** y **fácil de entender en el historial**.

**Revisión**: someter las refactorizaciones a code reviews para validar la calidad del diseño.


<h1 style="color:#66bb6a; font-size:36px;">Taxonomía</h1>

Uno de los mayores desafíos en el estudio de los *code smells* es su clasificación. Desde su introducción por Martin Fowler, numerosos investigadores han intentado organizar y categorizar estos olores de código.


<h2 style="color:#66bb6a; font-size:30px;">Falta de consenso</h2>

No existe una definición estándar ni una taxonomía única aceptada por toda la comunidad. Diferentes autores utilizan distintos criterios para agruparlos: por nivel de abstracción, tipo de problema que generan, o incluso por su localización en el código (intra o inter clase).

<h2 style="color:#66bb6a; font-size:30px;">Intentos de consensuar</h2>

Algunos trabajos clave, como la taxonomía de Mäntylä, han intentado sistematizar el estudio de los *code smells*. Otros, como Wake, Martin (Clean Code), y Atwood, han propuesto agrupamientos basados en la práctica. Más recientemente, se han incorporado enfoques empíricos que incluyen meta-análisis, categorización mediante atributos booleanos, y jerarquías de "bad smells".

<h2 style="color:#66bb6a; font-size:30px;">Situación actual</h2>

La falta de una fuente centralizada y estandarizada agrava el problema. A pesar de los esfuerzos por crear catálogos y sistemas de clasificación, los *code smells* siguen siendo tratados de forma dispersa y heterogénea tanto en la literatura científica como en la documentación técnica.

Esta fragmentación limita el avance en el estudio y uso práctico de los *code smells*, y refuerza la necesidad de catálogos unificados, como el propuesto en la tesis de Jerzyk (2022), que integren taxonomías existentes, sin perder la capacidad de expansión futura.

<h1 style="color:#66bb6a; font-size:36px;">Code smells</h1>
Se clasifican en los tipos ...
Veamos un ejemplo de cada uno de ellos.

<h1 style="color:#66bb6a; font-size:36px;">Functional Abusers</h1>

Son aquellos que van en contra de los principios de la programación funcional, lo que puede disminuir la fiabilidad del software.

Estos code smells surgen al no adherirse a conceptos clave como la ausencia de efectos secundarios y la inmutabilidad de los datos, o al utilizar construcciones imperativas en lugar de alternativas funcionales más claras.

<h2 style="color:#66bb6a; font-size:30px;">Mutable data</h2>

Surge cuando los datos pueden ser modificados después de su creación. Los datos mutables son considerados perjudiciales porque pueden provocar fallos inesperados en otras partes del código. Son una fuente importante de bugs difíciles de detectar ya que pueden ocurrir en condiciones poco comunes.

**Problemas principales**:
- **Propensos a errores**: Ya que pueden fallar inesperadamente en otras partes del código. Son una fuente importante de bugs difíciles de detectar porque pueden ocurrir bajo condiciones raras.
- **Difíciles de entender**: Es difícil razonar sobre variables que pueden cambiar en cualquier momento sin considerar todos los casos extremos.

**Soluciones comunes**:
- **Atributos inmutables**: Esto asegura que una vez que un objeto es creado, su estado no puede ser modificado.
- **Elegir un control de acceso apropiado**: Este método se refiere a restringir la visibilidad de los atributos y métodos de una clase para evitar modificaciones no deseadas desde fuera de la clase.
- **Separar consulta de modificador**: Este principio establece que un método debería realizar una acción (modificar el estado de un objeto) o devolver un valor (consultar el estado), pero no ambas cosas. 

<h3 style="color:#66bb6a; font-size:24px;">Código con Code Smell</h3>

```python
@dataclass
class Foo:
    name: str
    value: float
    premium: bool

# foo object instance will be passed around and modified
```

<h3 style="color:#66bb6a; font-size:24px;">Código sin Code Smell</h3>

```python
@dataclass(frozen=True)
class Foo:
    name: str
    value: float
    premium: bool

# foo object instance will be passed around and, if that is necessary,
# recreated, thus making the state change explicit for everyone and handled
```

<h1 style="color:#66bb6a; font-size:36px;">Lexical Abusers</h1>

Los Lexical Abusers se centran en el uso inapropiado o inefectivo de los elementos léxicos del código, principalmente los nombres (de variables, métodos, clases, etc.) y los comentarios. El problema principal es que estos elementos no transmiten el significado de manera clara o incluso pueden ser engañosos, inconsistentes o carecer de valor informativo, dificultando la comprensión del código a través del lenguaje utilizado.

<h2 style="color:#66bb6a; font-size:30px;">Inconsistent Name</h2>

Ocurre cuando se utilizan nombres diferentes para la misma funcionalidad en diferentes partes del código, lo que dificulta la comprensión y crea confusión.  La mente humana trabaja reconociendo patrones, por lo que la forma de operar y usar cada clase subsiguiente debería generalizarse a través del proyecto, facilitando y acelerando la comprensión del código.

**Problemas principales**:
- **Comprensibilidad**: La comunicación estandarizada a través de los nombres es vital para los atajos mentales.
- **Interrupción del estado de flujo**: El desarrollador espera un método dentro de una clase relacionada pero no lo encuentra y tiene que buscar variaciones sinónimas del método que desea.


**Soluciones comunes**:
- **Establecer una nomenclatura**: Una solución preventiva podría ser establecer y comunicar convenciones de nomenclatura claras para todo el equipo y realizar revisiones de código para asegurar su cumplimiento.

<h3 style="color:#66bb6a; font-size:24px;">Código con Code Smell</h3>

```python
class Human:
    def talk():
        ...

class Elf:
    def chat():
        ...
```

<h3 style="color:#66bb6a; font-size:24px;">Código sin Code Smell</h3>

```python
class Character(ABC):
    @abstractmethod
    def talk():
        """ Converse """

class Human(Character):
    def talk():
        ...

class Elf(Character):
    def talk():
        ...
```

<h1 style="color:#66bb6a; font-size:36px;">Obfuscators</h1>

Los Obfuscators abarcan Code Smells que reducen directamente la comprensibilidad del código a través de la complejidad innecesaria, la falta de claridad en el propósito o la inconsistencia en el estilo, yendo más allá de la simple elección de nombres. Los Obfuscators hacen que el código sea difícil de entender debido a su estructura, lógica o presentación, incluso si los nombres individuales pudieran ser comprensibles.

<h2 style="color:#66bb6a; font-size:30px;">Complicated Regex Expression</h2>

Son todas aquellas expresiones regulares que sin un previo contexto son difíciles de entender.

**Problemas principales**:
- **Evitar el uso innecesario para tareas simples**: Se considera una mala práctica utilizar Regex cuando el código estándar podría realizar la misma tarea de manera más clara y comprensible. 
- **La necesidad de explicaciones para expresiones complejas**: Incluso cuando el uso de Regex es apropiado, una expresión regular compleja y sin explicaciones puede dificultar significativamente su comprensión. Al igual que con clases y métodos, las expresiones regulares también deberían ser explicables, idealmente mediante el uso de variables con nombres descriptivos para construir la expresión.

**Soluciones comunes**:
- **Extraer Método**: Esta refactorización implica mover la expresión regular compleja a un método separado con un nombre que explique claramente su propósito.
- **Extraer Variable**: Esta técnica consiste en asignar partes de la expresión regular a variables con nombres descriptivos. Esto desglosa la complejidad de la Regex en fragmentos más pequeños y comprensibles, donde el nombre de cada variable aclara la función de la porción de la expresión regular que representa. 

<h3 style="color:#66bb6a; font-size:24px;">Código con Code Smell</h3>

```python
regex_pattern = '(\W|^)(\w*)\s-\s[0-9]?[0-9]:[0-9][0-9]'
```

<h3 style="color:#66bb6a; font-size:24px;">Código sin Code Smell</h3>

```python
def get_regex_pattern_current_city_time() -> str:
    """
    Example Match:
      - `Wroclaw - 17:42`
      - `Berlin - 17:42`
      - `San Jose - 10:42`

    Compiled: (\W|^)(\w*)\s-\s[0-9]?[0-9]:[0-9][0-9]
    """
    prevent_excessive_match = '(\W|^)'
    city = '(\w*)'
    indication = '\s-\s'
    hour = '[0-9]?[0-9]'
    minute = '[0-9][0-9]'
    time = f'{hour}:{minute}'
    return f"{prevent_excessive_match}{city}{indication}{time}"
```

<h1 style="color:#66bb6a; font-size:36px;">Dispensables</h1>

Son fragmentos de código que no aportan valor y podrían eliminarse sin afectar la funcionalidad. Hacen el código más largo, confuso o difícil de mantener.
<h2 style="color:#66bb6a; font-size:30px;">Speculative Generality</h2>

**Definición**: Crear clases, funciones o jerarquías innecesarias “por si acaso”. Esto va en contra del principio YAGNI (*You Ain’t Gonna Need It*).

**Problemas**:
- Aumenta la complejidad innecesaria del sistema
- Dificulta el mantenimiento y la comprensión del código. (You Ain’t Gonna Need It Principle Violation)

**Refactor Methods**:
- Collapse Hierarchy
- Inline Class / Inline Function
- Rename Method. (Sí es necesario)

<h3 style="color:#66bb6a; font-size:24px;">Código con Code Smell</h3>

```python

class Animal:
    health: int

class Human(Animal):
    name: str
    attack: int
    defense: int

class Swordsman(Human):
    ...

class Archer(Human):
    ...

class Pikeman(Human):
    ...
```

<h3 style="color:#66bb6a; font-size:24px;">Código sin Code Smell</h3>

```python
class Human:
    name: str
    health: int
    attack: int
    defense: int

class Swordsman(Human):
    ...

class Archer(Human):
    ...

class Pikeman(Human):
    ...
```

<h1 style="color:#66bb6a; font-size:36px;">Data Dealers</h1>
Se refiere a clases o métodos que actúan como meros contenedores de datos sin lógica, o que exponen demasiado sus datos internos, rompiendo el encapsulamiento.

<h2 style="color:#66bb6a; font-size:30px;">Message Chain</h2>

**Definición**: Acceso encadenado a datos a través de múltiples objetos. Viola el principio *Law of Demeter*.

**Problemas**:
- Fragilidad ante cambios en las clases intermedias
- Dificultad para el mantenimiento

**Refactor Methods**:
- Hide Delegate
- Extract Method
- Move Method



<h3 style="color:#66bb6a; font-size:24px;">Código con Code Smell</h3>

```python
class Minion:
    location: "Location"
    
    def action(self):
        ...
        if self.location.field.is_frontline():
            ...

class Location:
    field: "Field"

class Field:
    def is_frontline(self):
        ...
```


<h3 style="color:#66bb6a; font-size:24px;">Código sin Code Smell</h3>

```python
class Minion:
    location: Location
    def action(self):
        ...
        if self.is_frontline():
            ...
    
    def is_frontline(self):
        return self.location.is_frontline()

class Location:
    field: Field
    def is_frontline(self):
        return self.field.is_frontline()

class Field:
    def is_frontline(self):
        ...
```

<h1 style="color:#66bb6a; font-size:36px;">Object-Oriented Abusers</h1>

Mal uso o abuso de los principios de la programación orientada a objetos. 
Aparecen cuando una clase no está aprovechando adecuadamente características como la herencia, el encapsulamiento o el polimorfismo.

<h2 style="color:#66bb6a; font-size:30px;">Refused Bequest</h2>

**Definición**: Cuando una subclase hereda métodos o atributos que no necesita o no desea.

**Problemas**:
- Viola el principio de sustitución de Liskov
- Herencia inadecuada o confusa
> El principio de sustitución de Liskov describe que la relación entre subtipos y supertipos debe garantizar que cualquier propiedad demostrada sobre el objeto supertipo también sea válida para su objeto subtipo.

**Refactor Methods**:
- Extract Subclass
- Push Down Method / Field
- Replace Inheritance with Delegation
- Replace Inheritance with Delegation,
- Replace Superclass/Subclass with Delegate


<h3 style="color:#66bb6a; font-size:24px;">Código con Code Smell</h3>

```python

from abc import ABC, abstractmethod

class Minion(ABC):
    @abstractmethod
    def attack(self): ...

    @abstractmethod
    def move(self): ...

class Tower(Minion):
    def attack(self): ...
    def move(self):
        raise NotImplementedError
```


<h3 style="color:#66bb6a; font-size:24px;">Código sin Code Smell</h3>

```python

class Structure:
    def attack(self): ...

class Minion:
    def attack(self): ...
    def move(self): ...
```

<h1 style="color:#66bb6a; font-size:36px;">Couplers</h1>

Indican un **acoplamiento excesivo**, lo cual va en contra del diseño orientado a objetos.

Este grupo podría considerarse una **subcategoría de los Object-Oriented Abusers**, aunque los smells listados aquí se enfocan específicamente en problemas de alto acoplamiento.

<h2 style="color:#66bb6a; font-size:30px;">Feature Envy</h2>

Ocurre cuando un método dentro de un módulo manipula más campos o métodos de otro módulo que del suyo.

El comportamiento debe estar lo más cercano posible a los datos que utiliza.

Este olor indica una responsabilidad mal ubicada: el método está más acoplado a otro módulo que al suyo, lo que sugiere que probablemente debería ser movido.

**Problemas principales**:
- **Baja testeabilidad**: Dificulta la escritura de pruebas aisladas; requiere mocks.
- **Dificultad para reutilizar**: Los objetos quedan acoplados; extraer código sin arrastrar dependencias se vuelve complejo.
- **Problemas de bijección**: Es decir, hay una relación casi 1:1 forzada entre dos módulos o clases. Esto sugiere que una clase está demasiado acoplada a otra, al punto que parecen una sola unidad mal separada, lo que contradice la modularidad deseada.

**Soluciones comunes**:
- **Mover métodos/campos**: Si un método o atributo muestra mayor acoplamiento con una clase *A* que cohesión con su propia clase, debe ser movido a la clase *A*.

<h3 style="color:#66bb6a; font-size:24px;">Código con Code Smell</h3>

```python
from dataclasses import dataclass

@dataclass(frozen=True)
class ShoppingItem:
    name: str
    price: float
    tax: float

class Order:
    
    @staticmethod
    def get_bill_total(items: list[ShoppingItem]) -> float:
        return sum([item.price * item.tax for item in items])
    
    @staticmethod
    def get_receipt_string(items: list[ShoppingItem]) -> list[str]:
        return [f"{item.name}: {item.price * item.tax}$" for item in items]
    
    def create_receipt(self, items: list[ShoppingItem]) -> float:
        bill = self.get_bill_total(items)
        receipt = self.get_receipt_string(items).join('\n')
        return f"{receipt}\nBill {bill}"
```

<h3 style="color:#66bb6a; font-size:24px;">Código sin Code Smell</h3>

```python
@dataclass(frozen=True)
class ShoppingItem:
    name: str
    price: float
    tax: float
    
    @property
    def taxed_price(self) -> float:
        return self.price * self.tax
        
    def get_receipt_string(self) -> str:
        return f"{self.name}: {self.price * self.tax}$"

class Order:

    @staticmethod
    def get_bill_total(items: list[ShoppingItem]) -> float:
        return sum([item.taxed_price for item in items])

    @staticmethod
    def get_receipt_string(items: list[ShoppingItem]) -> list[str]:
        return [item.get_receipt_string() for item in items]
        
    def create_receipt(self, items: list[ShoppingItem]) -> float:
        bill = self.get_bill_total(items)
        receipt = self.get_receipt_string(items).join('\n')
        return f"{receipt}\nBill: {bill}$"
```

<h1 style="color:#66bb6a; font-size:36px;">Bloater</h1>

Se resume en la frase **“código que simplemente no necesita ser tan grande”**.

Se refiere a métodos o clases cuyo tamaño es desproporcionadamente grande en relación con lo que hacen, lo que los vuelve difíciles de mantener y entender.

**Empeoran** a medida que se insertan más líneas en funciones ya existentes al agregar nuevas funcionalidades.

<h2 style="color:#66bb6a; font-size:30px;">Combinatorial Explosion</h2>

La Combinatorial Explosion ocurre cuando una decisión que debería estar desacoplada (como el comportamiento de un objeto) se implementa mediante una **estructura rígida**.

**Problemas principales**:
- **Violación del principio DRY** (Don’t Repeat Yourself): Agregar una nueva funcionalidad implica duplicarla en varios lugares con mínimas diferencias.
- **Violación del principio Open/Closed** (OCP): La clase no está cerrada a modificaciones si constantemente “sugiere” al desarrollador que agregue otro elif.

**Soluciones comunes**:
- Remplazar valores de datos con objetos
- Remplazar condicionales de tipo con polimorfismos

<h3 style="color:#66bb6a; font-size:24px;">Código con Code Smell</h3>

```python
class Minion:
    name: str
    state: str = 'ready'
    def action(self):
        if self.state == 'ready':
            self.animate('standing')
        elif self.state == 'fighting':
            self.animate('fighting')
        elif self.state == 'resting':
            self.animate('resting')
            
    def next_state(self):
        if self.state == 'ready':
            return 'fighting'
        elif self.state == 'fighting':
            return 'resting'
        elif self.state == 'resting':
            return 'ready'
            
    def animate(self, animation: str):
        print(f"{self.name} is {animation}!")
```

<h3 style="color:#66bb6a; font-size:24px;">Código sin Code Smell</h3>

```python
from enum import Enum
from abc import ABC, abstractmethod

class State(ABC):
    @abstractmethod
    def next() -> "State": ...
        
    @abstractmethod
    def animate() -> str: ...

class Ready(State):
    def next():
        return States.FIGHT
        
    def animate():
        return 'standing'

class Fight(State):
    def next():
        return States.REST
        
    def animate():
        return 'fighting'

class Rest(State):
    def next():
        return States.READY
        
    def animate():
        return 'resting'


class States(Enum):
    READY: State = Ready
    FIGHT: State = Fight
    REST: State = Rest

class Minion:
    name: str
    state: State = States.Ready
    
    def action(self):
        self.state.animate()
        
    def next_state(self):
        self.state = self.state.next()
        
    def animate(self, animation: str):
        print(f"{self.name} is {self.state.animate()}!")

```

<h1 style="color:#66bb6a; font-size:36px;">Change Preventers</h1>

Los Change Preventers (prevenidores del cambio) son code smells que **rompen la relación biyectiva (uno a uno) entre clases y posibles cambios**. Es decir, para modificar algo se deben alterar muchas clases, o una clase se ve afectada por muchos tipos de cambios.

Esto es un problema importante en **entornos colaborativos**, donde trabajar en paralelo de forma independiente exige un buen encapsulamiento y separación de responsabilidades.

<h2 style="color:#66bb6a; font-size:30px;">Shotgun Surgery</h2>

Ocurre cuando una **responsabilidad** que podría estar cohesionada en un **solo componente**, se dispersa y termina siendo gestionada en **múltiples componentes**, generando el **"disparo en todas direcciones"** (de ahí el nombre Shotgun Surgery).

Esto genera un **acoplamiento** innecesario, ya que cada vez que se necesita hacer un cambio, hay que modificar varios componentes diferentes.

**Problemas principales**:
- Violación del Principio de Responsabilidad Única (SRP)
- El sistema pierde cohesión
- Aumenta la complejidad para mantener o cambiar el sistema

**Soluciones comunes**:
- Eliminar abstracciones innecesarias
- Extraer comportamiento repetido
- Combinar funciones
- Agrupar funciones relacionadas en una clase
- Mover métodos o campos donde realmente pertenecen

<h3 style="color:#66bb6a; font-size:24px;">Código con Code Smell</h3>

```python
def animate(animation: str): ...
def skip_turn(): ...

class Minion:
    energy: int
    def attack(self):
        if self.energy < 20:
            animate('no-energy')
            skip_turn()
            return
        ...
    
    def cast_spell(self):
        if self.energy < 50:
            animate('no-energy')
            skip_turn()
            return
        ...
    
    def block(self):
        if self.energy < 10:
            animate('no-energy')
            skip_turn()
            return
        ...
    
    def move(self):
        if self.energy < 35:
            animate('no-energy')
            skip_turn()
            return
        ...
```

<h3 style="color:#66bb6a; font-size:24px;">Código sin Code Smell</h3>

```python
class Minion:
    energy: int
    def attack(self):
        if not self.has_energy(20):
            return
        ...
    
    def cast_spell(self):
        if not self.has_energy(50):
            return
        ...
    
    def block(self):
        if not self.has_energy(10):
            return
        ...
    
    def move(self):
        if not self.has_energy(35):
            return
        ...
    
    def has_energy(self, energy_required: int) -> bool:
        if self.energy < energy_required:
            self.handle_no_energy()
            return False
        return True
    
    def handle_no_energy(self) -> None:
        animate('no-energy')
        skip_turn()
```
