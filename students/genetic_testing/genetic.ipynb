#### Origen de la idea

Los **Algoritmos Genéticos** se inspiran en la **evolución biológica**.

La idea central:
> "Equipo que gana no se toca"

*De muchas soluciones aleatorias, las mejores sobreviven y se combinan para generar mejores respuestas.*

*Supervivencia del más apto*

#### Hablemos un poco de Biologia

En biología, la **información genética** se transmite y evoluciona mediante:

- *ADN*: porta la información
- *Mitosis*: duplicación celular
- *Mutaciones*: errores/cambios que pueden mejorar la especie

> La evolución se da gracias a **selección**, **reproducción**, y **mutación**.

#### Algoritmo evolutivo

Una clase de algoritmos inspirados en lo anterior.
Tenemos un problema especifico → generamos muchas **soluciones candidatas** aleatorias.

Iteramos así:

1. Evaluamos cada solución con un **fitness** (qué tan buena es).
2. Seleccionamos las mejores.
3. Las reproducimos y modificamos.
4. Repetimos por N generaciones o hasta que la solucion me satisface.

Todos estos procesos tienen **parámetros configurables** con los que podemos ir jugando.

#### Evolución Biológica vs. Testing Genético

| Evolución Biológica       | Testing Genético                    |
|---------------------------|-------------------------------------|
| Individuos                | Casos de prueba / suites de test    |
| ADN / Genes               | Parámetros de entrada               |
| Fitness                   | Calidad del test                    |
| Reproducción              | Selección de mejores test cases     |
| Cruce (crossover)         | Combinar partes de test cases       |
| Mutación genética         | Cambiar parámetros del test         |
| Generaciones              | Iteraciones de optimización         |

*Así como la evolución mejora a los individuos, los GAs mejoran los tests.*


#### Test Adequacy Criteria

- Criterios que pueden servir como **función de fitness**..
  - **Coverage**: líneas, ramas, condiciones, caminos.
  - **Fault Detection**: encontrar bugs conocidos.
  - **Mutation Score**: qué tan bien mata versiones mutadas del código.


#### Casos de uso

- Casos de prueba manuales → lentos, limitados, sesgados.
- Espacio de entrada muy grande → imposible testear todos los inputs.
- GA permite:
  - Explorar soluciones no triviales.
  - Adaptarse a distintas funciones de fitness.
  - Automatizar generación y evolución de test cases.

#### Aplicaciones y Casos de Uso

- Pruebas de unidades matemáticas complejas.
- Testing de interfaces gráficas y flujos de interacción.
- Generación de pruebas para controladores embebidos.
- Aplicaciones en testing dirigido por modelos (MBT).
