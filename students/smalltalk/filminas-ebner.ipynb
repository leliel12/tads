# Mensajes


- Los objetos colaboran entre ellos a travez de mensajes. De hecho no existen operadores, solo mensajes.
- Los mensajes tienen 3 tipos **unarios**, **binarios** y **keywords**

En castellano, podemos sacar la siguiente informacion
- Un mensaje esta compuesto de un selector y opcionalmente argumentos
- Un mensaje siempre se le envia a un receptor, este puede ser un literal, un bloque, una variable o el resultado de evaluar otro mensaje.
- La combinacion de mensaje y receptor se llama *envio de mensaje*

# Mensajes
##  Mensajes unarios

- Los **mensajes unarios** son mensajes que no requieren ningun argumento. Son todos de la forma `reciver messageName`.
```smalltalk
89 sin
>>> 0.860069405812453

3 sqrt
>>> 1.732050807568877

Float pi
>>> 3.141592653589793

'blop' size
>>> 4

true not
>>> false
```


# Mensajes
## Mensajes binarios

- Los **mensajes binarios** son mensajes que requieren exactamente un argumento y su selector consiste en una secuencia de uno o mas caracteres del conjunto `{+, -, *, /, &, =, >, |, <, ~, @}`.

```smalltalk
100@100
>>> 100@100 "creates a Point object"

3 + 4
>>> 7

10 - 1
>>> 9

4 <= 3
>>> false

(4/3) * 3 == 4
>>> true "equality is just a binary message, and Fractions are exact"

(3/4) == (3/4)
>>> false "two equal Fractions are not the same object"
```

# Mensajes
## Mensajes keyword

- Los **mensajes keyword** son mensajes que requieren uno o mas argumentos y su selector consiste en una o mas *palabras clave* terminadas en dos puntos, es decir son de la forma `receiver selectorWordOne: argumentOne wordTwo: argumentTwo ... wordN: argumentN`
 

```smalltalk
2 between: 0 and: 10
>>> true
```
En el ejemplo anterior, el selector del mensaje es `between:and:`

```smalltalk
Color r:1 g:0 b:0
>>> Color red
```

A modo de ejemplo, en la sitaxis de Java el *mensaje enviado* anterior se corresponde a `Color.rgb(1,0,0)`


```smalltalk
1 to: 10
>>> (1 to: 10) "creates an interval"

| nums |
nums := Array newFrom: (1 to: 5).
nums at: 1 put: 6.
nums
>>> #(6 2 3 4 5)
```

# Mensajes
## Composicion y precedencia
Los 3 tipos de mensajes tienen distinta precedencia, lo que nos ayuda a poder componerlos de una manera simple y elegante.

A tener en cuenta:
1. Los mensajes *unarios* siempre se envian primero, luego los mensajes *binarios* y por ultimo los mensajes *keyword*.
2. Los mensajes en parentesis se envian primero a todo tipo de mensajes.
3. Los mensajes del mismo tipo se evaluan de izquierda a derecha.

Es decir podemos hacer cosas como
```smalltalk
1000 factorial / 999 factorial "en vez de (1000 factorial)/ (999 factorial)"
>>> 1000

2 raisedTo: 1 + 3 factorial "en vez de 2 (raisedTo: (1 + (3 factorial)))"
>>> 128
```

# Mensajes
## Composicion y precedencia

Esta bueno, pero hay que tener cuidado
```smalltalk
3 + 4 * 5
>>> 35 "(not 23) Binary messages sent from left to right"

3 + (4 * 5)
>>> 23

1 + 1/3
>>> (2/3) "and not 4/3"

1 + (1/3)
>>> (4/3)

1/3 + 2/3 " -> ((1/3) + 2) / 3"
>>> (7/9) "and not 1"

(1/3) + (2/3)
>>> 1
```

# Pseudovariables

En pharo hay 6 pseudovariables
1. `nil`
2. `true`
3. `false`
4. `self`
5. `super`
6. `thisContext`

Se llaman pseudovariables porque estan predefinidas y no se les puede asignar otro valor. 

Las variables `true`, `false`, y `nil` son constantes, mientras que `self`, `super` y `thisContext` varian dinamicamente mientras el codigo se ejecuta.

## Arrays
- Un array es una colección de tamaño **fijo** a la que se puede acceder mediante indices.
- La forma para acceder a los elementos del array es a travez del mensaje `at:` y `at:put:`.
 
**IMPORTANTISIMO:** los indices en Pharo empiezan en 1 💀☠️
<div style="max-width:1000px;margin: auto;">
    


</div>

- Podemos crear arrays de distintas formas
```smalltalk
#(4 4/6 'Pepe') size
>>> 3

Array with: 4 with: 'Francia' with: 3/2
>>> {4. 'Francia'. (3/2)}
```

# Sintaxis
## OrderedCollection
- Es una de las colecciones que puede expandirse.
- Ofrece mensajes como `add:"`, `addFirst:`, `addLast:` y `addAll:`
```smalltalk
| ordCol |
ordCol := OrderedCollection new.
ordCol add: 'C++'; add: 'Python'; addFirst: 'SmallTalk'.
ordCol
>>> an OrderedCollection('SmallTalk' 'C++' 'Python')
```
- Tambien podemos eliminar objetos

```smalltalk
ordCol add: 'Rust'.
ordCol remove: 'Rust'.
ordCol
>>> an OrderedCollection('SmallTalk' 'C' 'Python')
```
