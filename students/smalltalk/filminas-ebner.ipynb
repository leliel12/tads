# Sintaxis
## Mensajes


- Los objetos colaboran entre ellos a travez de mensajes. De hecho no existen operadores, solo mensajes.

- Los mensajes tienen 3 tipos **unarios**, **binarios** y **keywords**

- La precedencia de los mensajes es:
    1. Mensajes **unarios**
    2. Mensajes **binarios**
    3. Mensajes **keyword**
- Obviamente, los parentecis siempre son ejecutados primero

En castellano, podemos sacar la siguiente informacion
- Un mensaje esta compuesto de un selector y opcionalmente argumentos
- Un mensaje siempre se le envia a un receptor, este puede ser un literal, un bloque, una variable o el resultado de evaluar otro mensaje.
- La combinacion de mensaje y receptor se llama *envio de mensaje*

# Sintaxis
## Los 3 tipos de mensajes

- Los **mensajes unarios** son mensajes que son enviados a un objeto sin ningun tipo de informacion adicional
```
89 sin
>>> 0.860069405812453
3 sqrt
>>> 1.732050807568877
Float pi
>>> 3.141592653589793
'blop' size
>>> 4
true not
>>> false
Object class
>>> Object class "The class of Object is Object class (BANG)"
```


# Sintaxis
## Los 3 tipos de mensajes

- Los **mensajes binarios** son mensajes que consisten en operadores que ejecutan operaciones binarias basicas. son binarias porque involucran dos objetos, el receptor y el objeto argumento

`10 + 20` -> es un mensaje binario enviado al receptor 10 con el argumento 20.
