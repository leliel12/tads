# Ejemplo mas rebuscado: Parser Combinator



# Logica proposicional


```bash
<implies> ::= <or> '->' <implies>
            | <or>
    
<or> ::= <and> '|' <or>
       | <and>

<and> ::= <unary> '&' <and>
        | <unary>

<unary> ::= '!' <unary>
          | <ground>

<ground> ::= <scalar> | <ident>

<scalar> ::= '1' | '0'

<ident> ::= '[A-Za-z0-9]*'
```
