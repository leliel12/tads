# Termodinámica Clásica
## Ley 0
- El equilibrio térmico es una relación de equivalencia, y $T$ es una parametrización que indica si dos cuerpos pertenecen a la misma clase de equivalencia.
- En el contacto entre dos cuerpos a distinta $T$ se da un flujo de energía $Q$ llamado calor, del más caliente al más frío.


## 1° Ley
Conservación de la energía
$$\Delta U = Q - W $$
(punto de vista del sistema).

## Falta algo...
Ejemplos de procesos que conservan la energía pero no ocurren:
- Concentración de todo el aire en una esquina del aula.
- Cerveza en un vaso que se enfría y comienza a girar.
- Se enfría una punta del borrador y se calienta la otra.

Tiene que haber algo más que la conservación de la energía...

## Entropía
Existe una cantidad $S$ que mide la *dispersión de energía* a una temperatura dada
$$ dS = \frac{\delta Q_{rev}}{T} $$

Es útil porque nos permite calcular el mínimo derperdicio de energía ($Q$ liberado al ambiente), entre otras cosas.

## 2° Ley
Para todo proceso en un sistema cerrado 
$$ dS \geq 0 $$
donde la igualdad se da para procesos reversibles.

### Consecuencias
1. No existe la transferencia perfecta de $Q \rightarrow W$.
2. No se puede refrigerar de manera perfecta (debo inyectar $W$).
3. No existen los móviles perpetuos (la energía se “degrada”).
4. Todo proceso real aumenta la entropía (flecha del tiempo).
5. La entropía total del universo sólo puede mantenerse o aumentar (heat death).

### Termodinámica de la computación
Costo mínimo de entropía/calor para el cómputo.

## Macroestado vs Microestado
Usando el ejemplo de un gas

- Microestado: conjunto de valores de posición y velocidad para todas las partículas $\{(\vec{x_i}, \vec{v_i})\}_{i=1}^N$.

# Teoría de la Información
Pensando en cuantificar el contenido de información de un mensaje en un canal de comunicación, Claude Shannon definó la *entropía de información*.

Esta va a reflejar la información que gano si me dicen el microestado, sabiendo el macroestado.

## Motivación
- Dada una letra, adiviná la palabra.
- Intuitivamente, una “z” me da mucha más información que una “a”.
- La información que me da algo podría ser inversamente proporcional a su probabilidad, i.e., $I(L=l) \sim 1/p(L=l)$.
- Pero pero pero:
    - Evento seguro $\implies$ info. nula.
    - Evento imposible $\implies$ info. diverge.


¡$\log$ cumple esto!

Entonces, en realidad la info. que gano cuando me dicen que “salió $l$” debería ser proporcional a $$ I(L=l) = \log \left[ \frac{1}{p(L=l)} \right] $$


*A priori*, antes de realizar el experimento aleatorio, ¿cuál es la información que esperamos que nos aporte un caracter?

Debería ser el promedio de la información que aporta cada uno, pesado por su probabilidad $$ H(L) = \langle I(L) \rangle $$ $$ H(L) = \sum_{l=1}^W  p(L=l) \log \left[ \frac{1}{p(L=l)} \right] $$

donde $W$ es el número de eventos elementales del espacio muestral de la variable aleatoria $L$.

Casos límite:
- Si existe un evento seguro, i.e., con probabilidad 1, $H(L)=0$ ya que al saber el resultado de antemano, el caracter no aporta información.
- Se puede ver que el máximo de $H$ se da cuando todos los eventos son equiprobables, i.e., $p(L=l)=1/W$, donde tenemos que $$ H_{max}= \log W $$

Si medimos la información en $k_B/$nat, tenemos la entropía estadística $$S=k_B \log W $$


## Ejemplo: compresión de datos
- Transmisión de 4 caracteres “A”, “B”, “C” y “D” por un canal binario.
- Caracteres equiprobables: no se puede mejorar el encoding de 2 bits por caracter, e.g.,
    -  'A' como '00'
    -  'B' como '01'
    -  'C' como '10'
    -  'D' como '11'


Ahora digamos que:
- $p(X=A)=0.7$
- $p(X=B)=0.26$
- $p(X=C)=0.02$
- $p(X=D)=0.02$

Podemos asignar 'A' como '0', 'B' como '10', 'C' como '110', y 'D' como '111'.


````python
import numpy as np

# tomo log base 2 para que la unidad sea 'bits'
p_equip = 0.25
s_equip = - 4.0 * p_equip * np.log2(p_equip)
print(f"S_1 = {s_equip}")

p = [0.7, 0.26, 0.02, 0.02]
s = - np.sum(p * np.log2(p))
print(f"S_2 = {s}")
````

Dada la distribución de probabilidad sobre el espacio muestral (macroestado), nos hacían falta 2 bits para especificar el caracter (microestado) en el primer caso, mientras que en el segundo, **logramos hacerlo con sólo 1.09 bits** (en promedio).

# ¿Entropía del Software?
Intuición de los devs ([source](https://www.toptal.com/software/software-entropy-explained)):
> When software entropy overwhelms a project, *the project freezes*.
>
> Software entropy *cannot be reduced*.
>
> Software entropy arises from a *lack of knowledge*.
>
> Software entropy is the risk that changing existing software will result in *unexpected problems*, unmet objectives, or both.
>
> *Risk tolerance* and software entropy are related in that software entropy must be minimal to be certain that we will stay within a system’s risk tolerance during the next development iteration.

- Problema: se dicen muchas cosas que no cuadran, como si fuera una fuerza mágica que con el paso del tiempo aumenta irremediablemente “el desorden, el caos y la complejidad”.
- Sin embargo, con lo que vimos podemos “rescatar” esta intuición.

### Testing
- Consideramos programas en un único script de $N_{code}$ caracteres. Notar que esto incluye todos los programas de *hasta* ese tamaño (espacios en blanco o saltos de línea finales).
- Consideramos una cantidad numerable $N_{char}$ de caracteres válidos.
- Todos los programas posibles viven en un subconjunto del reticulado $\mathbb{Z}^{N_{code}}$, dado por tomar cada una de las coordenadas entre los valores permitidos $[1, N_{char}]$. Podemos llamar a este conjunto $$ \mathbb{Z}_{[1, N_{char}]}^{N_{code}} $$

- La gran mayoría de estos puntos no cumplen sintaxis. Llamemos al conjunto de los que sí, i.e., los válidos $$ \mathbb{V} \subset \mathbb{Z}_{[1, N_{char}]}^{N_{code}} $$
- Como *restricciones* sobre la semántica, vamos a pensar que tenemos un conjunto de proposiciones ${t_1, \dots, t_m}$ que son verdaderas cuando nuestros tests (unitarios o de integración) pasan. Es decir, son propiedades impuestas u observables del sistema como un todo. Definen nuestro *macroestado*.
- Podemos definir el conjunto de programas que pasan los tests $\mathbb{P} \subset \mathbb{V}$ como $$ \mathbb{P} = \{ x \in \mathbb{V} : (\land_{i=1}^m t_i)x = 1 \} $$ donde cada $t_i$ acepta un programa y devuelve un booleano.
- El *microestado* está dado por la secuencia de caracteres concreta $p \in \mathbb{P}$.

- Entonces, dado $\{t_i\}_{i=1}^m$, si $W$ es la cardinalidad de $\mathbb{P}$, en una primera aproximación (equiprobable, ergódica, o cota superior) la entropía de nuestro software es $$ S = k \log W $$
- Luego, si al hacer un cambio hacemos crecer el programa, efectivamente obtenemos un aumento de entropía: $$ \Delta N_{code} \geq 0 \implies \Delta W \geq 0 \implies \Delta S \geq 0 $$
- Como tener un mayor número de programas posibles que pasan mis tests incrementa la posibilidad de comportamiento imprevisto, mayor $S$ implica más probabilidad de consecuencias inesperadas.



- No es muy sorprendente si lo pensamos como entropía de información:
> S es la cantidad de información que gano sobre un programa si me lo muestran, i.e., si me dicen $p \in \mathbb{P}$, dado que yo ya sé que el programa pasa $\{t_i\}_{i=1}^m$.

- En el caso límite tenemos el sueño del TDD: tests tan restrictivos que haya un único programa que los pase, i.e., $W=1 \implies S=0$. Prácticamente podríamos deducir el programa de ver los tests que pasa. No habría comportamiento inesperado posible.

- **Hipótesis:** supongamos que existe una partición de $\mathbb{P}$ en $\mathbb{P_{+}}$ y $\mathbb{P_{-}}$, de cardinalidad respectiva $W_+$ y $W_-$, donde el primero contiene variaciones inofensivas del programa, y el segundo variaciones nocivas.
- Los elementos de $\mathbb{P_{-}}$ serían versiones buggeadas del código que pasan los tests.
- Los elementos de $\mathbb{P_{+}}$ podrían diferir en performance, dependencias, realizar chequeos extra, manejar los contextos de manera diferente, etc.
- Sea $f=W_{+}/W$ la fracción de programas admisibles, lo que *podría* pasar es que $$ f \xrightarrow[N_{code} \to \infty]{} 0 $$
- Esto implicaría que a medida que el código crece, es más difícil (improbable) llegar a $p \in \mathbb{P_{+}}$; “the project freezes”.

- ¿Cómo podemos combatir esto? Con más testing.
- Una manera de explorar $\mathbb{P}$ es hacer *mutation testing*.
- Los mutantes son elementos de $\mathbb{P}$. Está en nosotros decidir a qué partición pertenecen, y definir más tests para refinar $\mathbb{P}$ eliminando las porciones de $\mathbb{P_{-}}$ que identifiquemos.

*Podemos reducir la Entropía*. El código no es un sistema cerrado. Haciendo analogía con un sistema físico, podemos reducir su entropía *inyectando trabajo*.

La única entropía que sólo aumenta es la del universo, el único sistema verdaderamente cerrado.
