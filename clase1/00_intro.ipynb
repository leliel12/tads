# PROGRAMAR Vs DISEÑAR


| Aspecto | Programación | Diseño de Software |
|---------|--------------|---------------------|
| **Definición** | Implementación de soluciones mediante código | Planificación y estructuración de sistemas de software |
| **Alcance** | Enfocado en componentes individuales o módulos | Visión integral del sistema y su arquitectura |
| **Nivel de abstracción** | Bajo a medio (código, algoritmos) | Alto (arquitectura, patrones, sistemas) |
| **Enfoque principal** | Convertir requisitos en código funcional | Diseñar estructuras y sistemas escalables |
| **Habilidades clave** | Dominio de lenguajes, algoritmos, debugging | Modelado de sistemas, patrones de diseño, arquitectura |
| **Pensamiento** | Estructurado 🤖 | Lateral 🎨 |
| **Métricas de éxito** | Código funcional, eficiente, libre de errores | Sistemas escalables, mantenibles, adaptables |
| **Desafíos comunes** | Bugs, rendimiento, optimización | Complejidad, escalabilidad, adaptabilidad |
| **Responsabilidades** | Implementar funcionalidades específicas | Definir la estructura general del sistema |

----
> En nuestro caso no vamos a hacer una diferencia y vamos a tener una visión integradora


# Complementariedad

- El diseño de software proporciona el **plano** (blueprint) que la programación implementa 🤡
- El diseñador de software necesita entender las limitaciones de la programación 👍
- El programador debe comprender la visión general del diseño para implementarla correctamente 👍
- En equipos pequeños, a menudo la misma persona desempeña ambos roles 🤡
- El diseño evoluciona durante la programación a medida que se descubren nuevos desafíos 🍉 ➕ 🍷

# No existe la bala de plata (There is no silver bullet) 

> La esencia del software es la **complejidad**, la **conformidad**, la **mutabilidad** y la **invisibilidad 💣**.   
> La parte más difícil de construir un sistema de software es decidir qué construir. 

- **Complejidad**: Los sistemas de software tienen una cantidad enorme de estados posibles y múltiples interacciones entre componentes 🧉.
- **Conformidad**: El software debe adaptarse a sistemas existentes y requisitos externos que frecuentemente no siguen principios lógicos ☕ .
- **Mutabilidad**: El software se percibe como fácilmente modificable, lo que lleva a constantes cambios de requisitos. 🍵
- **Invisibilidad**: La estructura del software no es visible de manera inherente, lo que dificulta su representación y comprensión 😩.
