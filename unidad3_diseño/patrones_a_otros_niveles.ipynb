## Patrones de Arquitectura de Aplicación

- Son soluciones probadas para organizar la estructura interna de una aplicación, enfocándose en cómo los diferentes componentes se organizan, comunican y gestionan dentro de una única aplicación.
- Los patrones de arquitectura de aplicación no están consolidados en un único libro de referencia como ocurre con los patrones de diseño (GoF); y provienen de la evolución y práctica de la industria.

**Objetivos**

* La separación de responsabilidades
* La escalabilidad del código
* El testing
* La mantenibilidad

## Patrones de Arquitectura de Aplicación

#### Patrones de Presentación 🖥️

Organizan la lógica de presentación y la interfaz de usuario, separando responsabilidades para mejorar el mantenimiento y testing.

* MVC (Model-View-Controller) 
* MVP (Model-View-Presenter)
* MVVM (Model-View-ViewModel)
* ~PAC (Presentation-Abstraction-Control)~
* ~PM (Presentation Model)~


## Patrones de Arquitectura de Aplicación - (Model-View Patterns)

#### MVC (Model-View-Controller) 🎮

* **Model**: Lógica de negocio y datos
* **View**: UI (CSS, jQuery, HTML)
* **Controller**: Procesa requests y coordina Model-View

**Características**
* Separación clara de responsabilidades
* Controller actúa como mediador
* La View y el Model están desacoplados



## Patrones de Arquitectura de Aplicación - (Model-View Patterns)

#### MVP (Model-View-Presenter) 🎭
* **Model**: Similar a MVC
* **View**: UI completamente pasiva
* **Presenter**: Mediador View-Model sin manejo de tráfico

**Características**
* Relación uno a uno entre View y Presenter
* View no tiene referencia al Model
* Comunicación bidireccional View-Presenter
* View y Presenter se comunican por interfaces



## Patrones de Arquitectura de Aplicación - (Model-View Patterns)

#### MVVM (Model-View-ViewModel) 📱

* **Model**: Similar a MVC/MVP
* **View**: UI con binding automático
* **ViewModel**: Estado y comportamiento de la View

**Características**
* Relación muchos a uno entre View y ViewModel
* Binding bidireccional automático
* ViewModel no conoce la View
* Patrón Observer para notificaciones





## Patrones de Arquitectura de Aplicación - (Model-View Patterns)

---
*Puntos Clave de Diferenciación*:
- MVC: Controller maneja tráfico y coordina
- MVP: Presenter no maneja tráfico, solo mediación
- MVVM: Binding automático y relación muchos a uno

## Pipelines 🔄

* Divide el procesamiento en una serie de etapas secuenciales
* Cada etapa:
  - Recibe una entrada
  - Realiza una transformación
  - Pasa el resultado a la siguiente etapa

```
Entrada → [Etapa 1] → [Etapa 2] → [Etapa 3] → Salida
```

Es una composicion de funciones en su forma mas simple 

$$
    salida = (Etapa3(Etapa2(Etapa1(entrada)))
$$


#### Características Principales ⭐
* Cada etapa es independiente
* Procesamiento unidireccional
* Fácil de añadir/quitar etapas
* Separación clara de responsabilidades
* Puede ser síncrono o asíncrono

#### Casos de Uso 🎯
1. Procesamiento de imágenes: *Cargar → Redimensionar → Filtrar → Comprimir*
2. ETL (Extract, Transform, Load): *Extraer datos → Limpiar → Transformar → Cargar*
3. Compiladores: *Lexer → Parser → Optimizador → Generador de código*
4. Procesamiento de texto: *Tokenizar → Normalizar → Analizar → Formatear*

## Clasificación del Pipeline Pattern 🔄

#### 1. Como Patrón de Integración 🔗
* Principal categorización
* Usado para integrar sistemas y procesar datos
* Ejemplos:
  - ETL Pipelines
  - Message Processing Pipelines
  - Data Integration Flows

#### 2. Como Patrón Enterprise 🏢
* Cuando se usa a nivel de sistema completo
* Ejemplos:
  - Microservices Pipelines
  - CI/CD Pipelines
  - Data Processing Pipelines

#### 3. Como Patrón de Diseño 🎨
* Cuando se implementa a menor escala
* Ejemplos:
  - Builder Pattern Pipeline
  - Chain of Responsibility
  - Decorator chains

## Clasificación del Pipeline Pattern - Comparación de Contextos 📊

### Nivel Integración
* Enfoque en flujo de datos
* Conecta sistemas diferentes
* Maneja transformaciones
* Orientado a procesamiento

### Nivel Enterprise
* Enfoque en procesos de negocio
* Conecta componentes mayores
* Maneja orquestación
* Orientado a servicios

### Nivel Diseño
* Enfoque en objetos
* Conecta comportamientos
* Maneja responsabilidades
* Orientado a código

# Pipeline vs DAG 🔄

#### Pipeline 📝
* Flujo lineal y secuencial
* Una entrada → varios pasos → una salida
* Cada paso depende solo del anterior

```
A → B → C → D
```

#### DAG 🌳 (Grafo dirigido acíclico)
* Flujo no lineal pero direccional
* Múltiples caminos posibles
* Sin ciclos
* Un paso puede depender de varios anteriores

```
    A → B → D
    ↓   ↗ ↗
    C → E
```

#### Relación 🤝
* Un Pipeline es un caso especial de DAG
* Pipeline = DAG lineal
* DAG = Pipeline con ramificaciones
