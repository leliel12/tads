## Costo del diseño

- Ya vimos y calculamos el costo de la calidad.
- El verdadero costo del diseño no se limita a la fase inicial, sino que se extiende a lo largo de todo el ciclo de vida del software. Un diseño inadecuado puede llevar a costos mucho mayores en fases posteriores, mientras que un buen diseño inicial puede reducir significativamente los costos a largo plazo.
- El el principal desafío en el diseño e que no existe una solución única o "bala de plata" que pueda eliminar esta complejidad (*No silver bullets).
- Un buen diseño hace que un cambio conceptualmente facil, sea facil de realizar

## Diseño y rendimiento

- **La optimización prematura:** "...es la raíz de todo mal". Esto sugiere que el diseño inicial debería enfocarse en la claridad y la corrección, no en el rendimiento.
- **Ley de Amdahl:** Establece que la mejora en el rendimiento de un sistema está limitada por la parte del sistema que no puede ser mejorada.
- El rendimiento debe ser medido, no supuesto. El diseño debe permitir fácil instrumentación y **perfilado**.
- **Escalabilidad:** Un buen diseño debe considerar cómo el sistema se comportará con cargas de trabajo crecientes.
Patrones de diseño para rendimiento:
- A menudo, hay un **compromiso** entre un diseño limpio y modular y el máximo rendimiento posible.
- El rendimiento debe ser considerado desde el inicio como un **requisito no funcional** clave.

> Un buen diseño modular facilita la optimización selectiva del rendimiento al permitir cambios localizados en las partes críticas del sistema.


## KISS principle (Keep it simple, stupid)
---

- Usado ampliamente en las fuerzas armadas de USA, en el desarrollo de software.
- El termino fue acuñado por el Ing. Kelly Johnson. Refiriendose a al desafío de que el avión a reacción que estaban diseñando debe ser reparable por un mecánico promedio en  condiciones de combate con solo unas pocas herramientas. 
- Se refiere a la relación entre la forma en que se rompen las cosas y la sofisticación disponible para repararlas.




## DRY - Don't repeat yourself

- Toda pieza de conocimiento debe tener solamente una representación autoritaria e inequívoca dentro de un sistema.
- O en cristiano: No hacer dos funciones que hacen lo mismo.
- Formulado por  Andy Hunt y Dave Thomas en el libro *The Pragmatic Programmer*



## Worse is better

- Acuñado por Richard P. Gabriel para describir la dinámica de la aceptación del software.
- Es la idea **subjetiva** de que la calidad no necesariamente aumenta con la funcionalidad
- que hay un punto en el que menos funcionalidad ("peor") es una opción preferible ("mejor") en términos de practicidad y usabilidad. 
- El software que es limitado, pero fácil de usar, puede ser más atractivo para el usuario y el mercado.

## Worse is better

"Worse-is-Better" es un modelo de diseño e implementación de software que tiene las siguientes características (en un orden de importancia aproximadamente descendente):

- **Sencillez (Simplicity)**
    El diseño debe ser simple, tanto en la implementación como en la interfaz. 

- **Exactitud (Correctness)**
    El diseño debe ser correcto en todos los aspectos observables, pero es un poco mejor ser simple que correcto.

- **Consistencia (Consistency)**
    El diseño no debe ser demasiado inconsistente.

- **Completitud (Completeness)**
    El diseño debe cubrir tantas situaciones importantes como sea práctico. 

## "You Aren't Gonna Need It" (No lo vas a necesitar, YAGNI)

- Es un principio de desarrollo de software que forma parte de la metodología de Programación Extrema (XP) y se relaciona con el diseño simple y eficiente del software. 

- Es más eficiente y menos costoso agregar funcionalidades cuando realmente se necesitan, en lugar de intentar anticipar todas las posibles necesidades futuras.

**Los puntos clave de YAGNI son:**

1. Evitar agregar funcionalidad hasta que sea necesaria
2. Enfocarse en los requisitos actuales, no en los potenciales futuros
3. Reducir la complejidad del código
4. Ahorrar tiempo y recursos de desarrollo
