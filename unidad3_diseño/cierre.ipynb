# KISS, DRY y YAGNI: Los tres mosqueteros del "No te hagas la compliques"

- **KISS (Keep It Simple, Stupid):** Este es el amigo minimalista que llega a la reunión con una camiseta blanca y  te mira con desprecio cuando intentas meter un framework de 500 dependencias para hacer un “Hola, mundo”. 
- **DRY (Don’t Repeat Yourself):** ;antiene el código limpio, pero si te pasas de rosca, terminas con abstracciones tan enredadas que ni el mismísimo Einstein podría descifrarlas.
- **YAGNI (You Ain’t Gonna Need It):** Te frena cuando estás a punto de meter un sistema de autenticación biométrica para una app que solo muestra memes. Es el antídoto contra los programadores que sueñan con “escalabilidad infinita” mientras su app la usan tres gatos. Pero ojo, si lo escuchas demasiado, terminas con un sistema tan pelado que no aguanta ni una brisa.

> KISS es el que cocina arroz blanco porque “es suficiente”, DRY es el que te grita por usar dos recetas distintas para el mismo arroz, y YAGNI es el que te dice que no necesitas arroz porque “con pan ya estás”. Juntos son como un trío de chefs minimalistas que te salvan de morir envenenado por tu propia ambición culinaria.

# Heurísticas de Diseño Orientado a Objetos: Los gurús espirituales del código

- **Inmutabilidad:** Los objetos inmutables son como estatuas: no cambian aunque les tires pintura. Perfectos para evitar que algún idiota rompa todo accidentalmente.
- **Lenguaje Ubicuo y Metáforas del Dominio:** El poeta del equipo que insiste en que todos hablemos el mismo idioma. “Si el cliente dice ‘carrito de compras’, no lo llames ‘cesta de transacciones’, ¡pedazo de snob!”
- **Null Object Pattern:** “Tranquilo, no hay nada, pero no te voy a hacer crash”. Es como un airbag para tu código.
- **Arquitectura del Sistema en el Código:** El arquitecto que quiere que el código refleje el sistema como si fuera un plano de AutoCAD. Si el negocio es un supermercado, el código debe oler a pasillo de detergentes, no a galaxia abstracta.
- **Entidades del Dominio como Objetos:** El que convierte conceptos del negocio en objetos con vida propia. “¿Un cliente? ¡Eso no es un string, es un ser con nombre, dirección y un historial de quejas porque le llegó el pedido tarde!”.
- **Invarianza de Objetos:** Es como un padre que no deja salir a su hijo sin que se peine primero.
- **Encapsulamiento:** El paranoico que esconde todo detrás de una pared.
- **Objetos Cohesivos y Completos:** El perfeccionista que dice: “Si vas a hacer algo, hazlo bien o no lo hagas”. 

> Estos son como un grupo de gurús en una comuna, pero en lugar de meditar, están obsesionados con que el código sea puro, claro y no se rompa. Inmutabilidad es el que no cambia ni aunque le ruegues, Null Object es el que te da un placebo para que no llores, y Encapsulamiento es el que tiene un búnker con candado para sus secretos.

# Clean Architecture, Clean Code, y sus compinches: Los maniáticos de la limpieza

- **Clean Architecture:**  “Negocio aquí, datos allá, y la UI en la cima, ¡pero que no se toquen, carajo!”. Es el que te salva de que un cambio en la base de datos te obligue a reescribir toda la app.
- **Clean Code Principles:**  “Nombres claros, funciones cortas, y si tu método tiene más de 10 líneas, te doy con la regla”. Es el que hace que tu código sea legible hasta para tu abuela.
- **Boy Scout Rule:**  “¿Este método está feo? Lo refactorizo aunque no sea mi tarea”.
- **Command Query Separation (CQS):** ¡Si preguntas, no modifiques! ¡Si modificas, no preguntes!”.
- **Dependency Rule:** El que pone orden en las dependencias como un guardia de tráfico. “Todo apunta hacia adentro, nada hacia afuera, ¡o se arma el caos!”.
- **Humble Object Pattern:** “Yo solo hago lo mínimo, el resto lo dejo a los cracks”. 
- **Transformation Priority Premise:** El estratega que te dice cómo transformar el código paso a paso sin romper todo. Es como un chef que te enseña a hacer un pastel sin que se te queme la cocina.
- **Four Rules of Simple Design:** Los cuatro mandamientos del código simple: pasa los tests, no repite, es claro y es minimalista. Es como un decálogo para no escribir código que parezca un chiste malo.
- **STABLE Patterns:** Los que te dan estabilidad como un edificio antisísmico. Son los patrones que aseguran que tu sistema no colapse cuando le metesun cambio.

> Clean Architecture es como Marie Kondo ordenando tu clóset: “Si no trae alegría, ¡a la basura!”. Clean Code es tu mamá diciéndote que laves los platos antes de que se acumulen. Boy Scout Rule es el amigo que recoge la basura después de la fiesta, y SRP es el que te dice que no invites a 50 personas si solo tienes lugar para 10. Juntos son como un equipo de limpieza obsesionado con que tu código no huela a podrido.

# SOLID, CUPID y GRASP: Los filósofos del diseño

- **SOLID:** El quinteto de rockstars del diseño orientado a objetos. Single Responsibility es el vocalista que no quiere compartir el escenario. Open/Closed es el que te deja añadir sin romper nada. Liskov Substitution es el que asegura que los hijos no deshonren a los padres. Interface Segregation es el que dice: “No me hagas implementar basura que no uso”. Dependency Inversion es el que pone las dependencias patas arriba para que todo fluya.
- **CUPID:** El primo moderno de SOLID, más relajado y pragmático. Es como SOLID pero con una cerveza en la mano, diciendo: “Hagamos código que sea alegre, no solo perfecto”. Sus principios (como ser composable, unix-like, predecible) son como un manual para no ser un programador aburrido.
- **GRASP:** El abuelo sabio que te da consejos generales. “Acoplamiento bajo, cohesión alta, y no dejes que los objetos se metan en la vida de los demás”. Es como un manual de etiqueta para que tus clases no sean unas chismosas entrometidas.

> SOLID es como una banda de rock clásica que todos respetan pero a veces suena anticuada. CUPID es la banda indie que toca en bares pequeños y se siente más cool. GRASP es el abuelo que te cuenta historias de los 70 pero todavía tiene razón en todo.
